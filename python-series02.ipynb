# Python Series 02 : Sinusoïdes
Ce cahier montre comment générer et tracer des signaux sinusoïdaux en Python. 
Il inclut la théorie, la génération numérique, et la visualisation.

Une onde sinusoïdale est définie par :

s(t) = A * sin(2π f t + φ)

où :
- A est l'amplitude,
- f la fréquence en Hz,
- φ la phase en radians,
- t le temps.

```python
import numpy as np       # Pour les calculs numériques et les tableaux
import matplotlib.pyplot as plt   # Pour tracer les graphiques
```

```python
t = np.linspace(0, 1, 1000)  
```

```python
s = np.sin(2 * np.pi * 5 * t)
```

```python
plt.plot(t, s)
plt.title("Signal sinusoïdal")
plt.xlabel("Temps [s]")
plt.ylabel("Amplitude")
plt.grid(True)
plt.savefig("sine_signal.png")
plt.show()
```

```python
s2 = np.cos(2 * np.pi * 5 * t)
```

```python
fig, ax1 = plt.subplots()
ax1.plot(t, s, label="sin")
ax2 = ax1.twinx()   # Second axe y
ax2.plot(t, s2, 'r', label="cos")
ax1.set_xlabel("Temps [s]")
ax1.set_ylabel("Amplitude sin", color='b')
ax2.set_ylabel("Amplitude cos", color='r')
fig.legend(loc="upper right")
plt.grid(True)
plt.savefig("sine_cos_signal.pdf")
plt.show()
```

```python
def sinusoid(frequency, amplitude, phase, t):
    """
    Génère un signal sinusoïdal.

    Paramètres:
    - frequency : fréquence en Hz
    - amplitude : amplitude du signal
    - phase : déphasage en radians
    - t : array numpy des temps

    Retour:
    - signal numpy array
    """
    return amplitude * np.sin(2 * np.pi * frequency * t + phase)
```

```python
s1 = sinusoid(5, 1, 0, t)           # 5 Hz, amplitude 1, phase 0
s2 = sinusoid(10, 0.5, np.pi/4, t)  # 10 Hz, amplitude 0.5, phase π/4

plt.plot(t, s1, label="5 Hz")
plt.plot(t, s2, label="10 Hz")
plt.xlabel("Temps [s]")
plt.ylabel("Amplitude")
plt.title("Deux signaux sinusoïdaux")
plt.legend()
plt.grid(True)
plt.show()
```
