## 1.1 Let's begin with the draft...

# Sinusoids, from math to code

**Goal:** Generate a simple sinusoidal signal (sine wave), plot it, and save the figure.  
We will use **NumPy** to generate the data and **Matplotlib** to plot the curve.

## Mathematical background

A sinusoid is defined as:

\[
s(t) = A \cdot \sin(2 \pi f(t) + \varphi)
\]

where:
- \( A \) = amplitude
- \( f \) = frequency (Hz)
- \( t \) = time (s)
- \( \varphi \) = phase (rad)

For this draft version, we will use:
- Amplitude \( A = 1 \)
- Frequency \( f = 1 \) Hz
- Phase \( \varphi = 0 \)


```python
import numpy as np
import matplotlib.pyplot as plt

# Generate the time array: 1000 values between 0 and 1 second
t = np.linspace(0, 1, 1000)

# Generate the sine wave: s(t) = sin(2πt)
s = np.sin(2 * np.pi * t)

# Plot
plt.figure(figsize=(8,4))
plt.plot(t, s, label="sin(2πt)")
plt.title("Sine wave (1 Hz)")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.grid(True)
plt.legend()
plt.show()

# Save as PNG
plt.figure(figsize=(8,4))
plt.plot(t, s, label="sin(2πt)")
plt.title("Sine wave (1 Hz)")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.grid(True)
plt.legend()
plt.savefig("sinusoid.png", dpi=300)
plt.close()
```

## 1.2 ...and now the better version.

## Improved version

Now we add:
- A cosine function on the same figure
- A legend
- A second y-axis
- Export as PDF


```python
# Generate sine and cosine
sine = np.sin(2 * np.pi * t)
cosine = np.cos(2 * np.pi * t)

# Create figure
fig, ax1 = plt.subplots(figsize=(8,4))

# First y-axis for sine
ax1.plot(t, sine, 'b-', label="sin(2πt)")
ax1.set_xlabel("Time (s)")
ax1.set_ylabel("Sine amplitude", color='b')
ax1.tick_params(axis='y', labelcolor='b')
ax1.grid(True)

# Second y-axis for cosine
ax2 = ax1.twinx()
ax2.plot(t, cosine, 'r-', label="cos(2πt)")
ax2.set_ylabel("Cosine amplitude", color='r')
ax2.tick_params(axis='y', labelcolor='r')

# Add legend
fig.legend(loc="upper right", bbox_to_anchor=(1,1), bbox_transform=ax1.transAxes)

# Show and save
plt.title("Sine and Cosine")
plt.show()
fig.savefig("sine_cosine.pdf")
plt.close(fig)
```

## 1.3 Create a notebook to plot sinusoidal signals (final version).

We now create a function `sinusoid` that generates sinusoidal signals.  
This allows code reusability and clarity.  
We also improve variable names, add comments (why, not what), and docstrings.


```python
def sinusoid(frequency: float, amplitude: float, phase: float, duration: float, sampling_rate: int = 1000):
    """
    Generate a sinusoidal signal.

    Parameters
    ----------
    frequency : float
        Frequency of the sinusoid (Hz)
    amplitude : float
        Amplitude of the sinusoid
    phase : float
        Initial phase (rad)
    duration : float
        Signal duration (s)
    sampling_rate : int
        Number of samples per second

    Returns
    -------
    time_array : numpy.ndarray
        Time values
    signal : numpy.ndarray
        Sinusoidal signal
    """
    time_array = np.linspace(0, duration, int(duration * sampling_rate))
    signal = amplitude * np.sin(2 * np.pi * frequency * time_array + phase)
    return time_array, signal

# Example: two signals with different frequencies
t1, s1 = sinusoid(frequency=1, amplitude=1, phase=0, duration=1)
t2, s2 = sinusoid(frequency=1, amplitude=1, phase=1, duration=1)

# Plot
plt.figure(figsize=(8,4))
plt.plot(t1, s1, label="1 Hz sine")
plt.plot(t2, s2, label="5 Hz sine")
plt.title("Two sinusoidal signals")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.grid(True)
plt.legend()
plt.show()
```
