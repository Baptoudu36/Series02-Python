## Mathematical background

A sinusoid is defined as:

\[
s(t) = A \cdot \sin(2 \pi f t + \varphi)
\]

where:
- \( A \) = amplitude
- \( f \) = frequency (Hz)
- \( t \) = time (s)
- \( \varphi \) = phase (rad)

For this draft version, we will use:
- Amplitude \( A = 1 \)
- Frequency \( f = 1 \) Hz
- Phase \( \varphi = 0 \)


```python
import numpy as np
import matplotlib.pyplot as plt

# Generate the time array: 1000 values between 0 and 1 second
t = np.linspace(0, 1, 1000)

# Generate the sine wave: s(t) = sin(2πt)
s = np.sin(2 * np.pi * t)

# Plot
plt.figure(figsize=(8,4))
plt.plot(t, s, label="sin(2πt)")
plt.title("Sine wave (1 Hz)")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.grid(True)
plt.legend()
plt.show()

# Save as PNG
plt.figure(figsize=(8,4))
plt.plot(t, s, label="sin(2πt)")
plt.title("Sine wave (1 Hz)")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.grid(True)
plt.legend()
plt.savefig("sinusoid.png", dpi=300)
plt.close()
```

```python
def sinusoid(frequency, amplitude, phase, t):
    """
    Génère un signal sinusoïdal.

    Paramètres:
    - frequency : fréquence en Hz
    - amplitude : amplitude du signal
    - phase : déphasage en radians
    - t : array numpy des temps

    Retour:
    - signal numpy array
    """
    return amplitude * np.sin(2 * np.pi * frequency * t + phase)
```

```python
s1 = sinusoid(5, 1, 0, t)           # 5 Hz, amplitude 1, phase 0
s2 = sinusoid(10, 0.5, np.pi/4, t)  # 10 Hz, amplitude 0.5, phase π/4

plt.plot(t, s1, label="5 Hz")
plt.plot(t, s2, label="10 Hz")
plt.xlabel("Temps [s]")
plt.ylabel("Amplitude")
plt.title("Deux signaux sinusoïdaux")
plt.legend()
plt.grid(True)
plt.show()
```
