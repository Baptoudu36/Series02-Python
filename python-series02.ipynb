# Sinusoids, from math to code

**Goal:** Generate a simple sinusoidal signal (sine wave), plot it, and save the figure.  
We will use **NumPy** to generate the data and **Matplotlib** to plot the curve.

## Mathematical background

A sinusoid is defined as:

\[
s(t) = A \cdot \sin(2 \pi f(t) + \varphi)
\]

where:
- \( A \) = amplitude
- \( f \) = frequency (Hz)
- \( t \) = time (s)
- \( \varphi \) = phase (rad)

For this draft version, we will use:
- Amplitude \( A = 1 \)
- Frequency \( f = 1 \) Hz
- Phase \( \varphi = 0 \)


```python
import numpy as np
import matplotlib.pyplot as plt

# Generate the time array: 1000 values between 0 and 1 second
t = np.linspace(0, 1, 1000)

# Generate the sine wave: s(t) = sin(2πt)
s = np.sin(2 * np.pi * t)

# Plot
plt.figure(figsize=(8,4))
plt.plot(t, s, label="sin(2πt)")
plt.title("Sine wave (1 Hz)")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.grid(True)
plt.legend()
plt.show()

# Save as a PNG file
plt.figure(figsize=(8,4))
plt.plot(t, s, label="sin(2πt)")
plt.title("Sine wave (1 Hz)")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.grid(True)
plt.legend()
plt.savefig("sinusoid.png", dpi=300)
plt.close()
```

## Improved version

Now we add:
- A cosine function on the same figure
- A legend
- A second y-axis
- Export as PDF


```python
def sinusoid(frequency: float, amplitude: float, phase: float, duration: float, sampling_rate: int = 1000):
    """
    Generate a sinusoidal signal.

    Parameters
    ----------
    frequency : float
        Frequency of the sinusoid (Hz)
    amplitude : float
        Amplitude of the sinusoid
    phase : float
        Initial phase (rad)
    duration : float
        Signal duration (s)
    sampling_rate : int
        Number of samples per second

    Returns
    -------
    time_array : numpy.ndarray
        Time values
    signal : numpy.ndarray
        Sinusoidal signal
    """
    time_array = np.linspace(0, duration, int(duration * sampling_rate)) # Creates evenly spaced time points from 0 to duration.
    signal = amplitude * np.sin(2 * np.pi * frequency * time_array + phase) # Computes the sine value at each time point.
    return time_array, signal # Used for plottiing.

# Properties of the two signals with their units : Hz, radians, seconds & Hz
t1, s1 = sinusoid(frequency=1, amplitude=1, phase=0, duration=1) 
t2, s2 = sinusoid(frequency=2, amplitude=1, phase=0, duration=1) 

# Plot the figures
plt.figure(figsize=(8,4))
plt.plot(t1, s1, label="Sinus 1")
plt.plot(t2, s2, label="Sinus 2")
plt.title("Two sinusoidal signals")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.grid(True)
plt.legend()
plt.show()
```

## 1.3 Create a notebook to plot sinusoidal signals (final version).

We now create a function `sinusoid` that generates sinusoidal signals.
